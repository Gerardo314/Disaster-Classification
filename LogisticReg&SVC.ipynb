Making the test/ train

```python
X_train = df_train["text"]
y_train = df_train["category"]

X_test = df_test["text"]
y_test = df_test["category"]
```

```python
model= Pipeline([
    ("tfidf",TfidfVectorizer(
        max_features=15000,
        ngram_range= (1,2),
        min_df =3,
        stop_words="english"
    )),
    ("logreg",LogisticRegression(
        max_iter=1000,
        n_jobs=-1,
        multi_class="auto"
    ))
])
```

```python
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
model2= Pipeline([
    ("tfidf",TfidfVectorizer(
        max_features=15000,
        ngram_range= (1,2),
        min_df =3,
        stop_words="english"
    )),
    ("logreg",LogisticRegression(
        max_iter=1000,
        n_jobs=-1,
        multi_class="auto",
        class_weight="balanced"
    ))
])
```

```python
model2.fit(X_train, y_train)
y_pred = model2.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
model3= Pipeline([
    ("tfidf",TfidfVectorizer(
        max_features=15000,
        ngram_range= (1,2),
        min_df =3,
    
    )),
    ("logreg",LogisticRegression(
        max_iter=1000,
        n_jobs=-1,
        multi_class="auto",
        class_weight="balanced"
    ))
])
```

```python
model3.fit(X_train, y_train)
y_pred = model3.predict(X_test)
print(classification_report(y_test, y_pred))
```

Using SVM

```python
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer

model_svm = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=15000,
        ngram_range=(1,2),
        min_df=2
    )),
    ("svm", LinearSVC())
])

model_svm.fit(X_train, y_train)
y_pred = model_svm.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer

model_svm2 = Pipeline([
    ("tfidf", TfidfVectorizer(
        max_features=15000,
        ngram_range=(1,2),
        min_df=2
    )),
    ("svm", LinearSVC(class_weight="balanced"))
])

model_svm2.fit(X_train, y_train)
y_pred = model_svm2.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
for C in [0.5, 1.0, 2.0]:
    print("\n=== C =", C, "===\n")
    model_tmp = Pipeline([
        ("tfidf", TfidfVectorizer(
            max_features=15000,
            ngram_range=(1, 2),
            min_df=2
        )),
        ("svm", LinearSVC(class_weight="balanced", C=C))
    ])

    model_tmp.fit(X_train, y_train)
    y_pred = model_tmp.predict(X_test)
    print(classification_report(y_test, y_pred))
```

Best Model is the SVC with C=0.5, 15000 max fatures, min_df =2

```python
model_svc = Pipeline([
        ("tfidf", TfidfVectorizer(
            max_features=15000,
            ngram_range=(1, 2),
            min_df=2
        )),
        ("svm", LinearSVC(class_weight="balanced", C=0.5))
    ])

model_svc.fit(X_train, y_train)
y_pred = model_svc.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
with open("SVC_model.pkl", "wb") as f:
    pickle.dump(model_svc, f)
```
